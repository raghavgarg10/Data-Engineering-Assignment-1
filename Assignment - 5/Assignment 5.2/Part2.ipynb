# **Q1. What are the key limitations of traditional MapReduce that make Spark a preferred choice for modern big data processing?**

**Answer:**

Traditional MapReduce was one of the first frameworks used for large-scale data processing, but it has several limitations that affect performance and efficiency. These limitations led to the development and adoption of Apache Spark.

**Limitations of MapReduce**

**1. Disk-Based Processing**

MapReduce writes intermediate results to disk after each processing stage.
Frequent disk read/write operations increase execution time.

**2. Slow Performance**

Due to heavy disk I/O, MapReduce is slower when processing large datasets.
Performance decreases significantly for complex workflows.

**3. Inefficient for Iterative Algorithms**

Machine Learning and Graph Processing algorithms require repeated access to the same data.
MapReduce reloads data from disk in every iteration, making such tasks inefficient.

**4. Complex Development Process**

Developers need to write separate Map and Reduce functions.
Code becomes lengthy and difficult to maintain.

**5. High Latency**

MapReduce is designed mainly for batch processing.
It is not suitable for real-time analytics or interactive queries.

**Why Spark is Preferred**

Apache Spark overcomes these limitations through:

- **In-Memory Processing** – stores intermediate data in RAM instead of disk.
- **Faster Execution** – can be significantly faster than MapReduce for many workloads.
- **Ease of Use** – provides simple APIs in Python, Scala, Java, and R.
- **Support for Advanced Analytics** – includes libraries for SQL, Machine Learning, Streaming, and Graph Processing.
- **Better Support for Iterative Workloads** – ideal for machine learning and data science applications.

# **Q2. Explain how Spark uses In-Memory Computing to speed up iterative machine learning algorithms compared to disk-based systems.**

**Answer:**

Apache Spark uses **In-Memory Computing**, which means it keeps intermediate data in RAM instead of storing it on disk after every operation.

**How it works:**

1. Data is loaded into memory only once.
2. Intermediate results can be cached and reused whenever required.
3. Machine learning algorithms can access the same data repeatedly without reading it from disk again.
4. This reduces processing time and improves performance.

**Comparison with Disk-Based Systems**


| Spark (In-Memory) | Disk-Based Systems |
|-------------------|-------------------|
| Stores intermediate data in RAM | Stores intermediate data on disk |
| Faster execution | Slower because of disk I/O |
| Suitable for iterative processing | Less efficient for repeated computations |
| Lower latency | Higher latency |

**Conclusion**

Since Spark keeps frequently used data in memory, it can perform iterative machine learning tasks much faster than traditional disk-based systems. This makes Spark a popular choice for modern big data processing.

# **Q4. Given a DataFrame df_sales, write a query to filter for rows where the region is 'West' and then group by product_category to find the average sale_amount.**

**Answer:**

This query filters the records where the `region` is **'West'** and then groups the data by `product_category` to calculate the average `sale_amount`.

```python
# Import required function
from pyspark.sql.functions import avg

# Sample Sales Data
data = [
    ("West", "Books", 450),
    ("West", "Books", 650),
    ("South", "Furniture", 1200),
    ("West", "Furniture", 1500),
    ("East", "Books", 700),
    ("West", "Sports", 900)
]

# Column names
columns = ["region", "product_category", "sale_amount"]

# Create DataFrame
df_sales = spark.createDataFrame(data, columns)

print("Original Sales Data:")
df_sales.show()

# Filter records for West region
west_sales = df_sales.filter(
    df_sales.region == "West"
)

# Calculate average sale amount by product category
result = west_sales.groupBy(
    "product_category"
).agg(
    avg("sale_amount").alias("Average_Sale_Amount")
)

print("Average Sale Amount for West Region:")
result.show()
```

# **Q5. What is the difference between .na.drop() and .na.fill()? Provide a code example of filling null values in a status column with the string 'Unknown'.**

### **.na.drop()**

- Removes rows that contain null values.
- Useful when incomplete records are not required.

### **.na.fill()**

- Replaces null values with a specified value.
- Helps retain all records while handling missing data.

### **Difference**

| .na.drop() | .na.fill() |
|------------|------------|
| Removes rows with null values | Replaces null values with a given value |
| Reduces the number of records | Keeps all records intact |
| Used when missing data should be discarded | Used when missing data can be replaced |

```python
# Sample Data
data = [
    (101, "Active"),
    (102, None),
    (103, "Inactive"),
    (104, None),
    (105, "Pending")
]

# Column names
columns = ["employee_id", "status"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

print("Original Data:")
df.show()

# Replace null values with 'Unknown'
df_filled = df.na.fill(
    {"status": "Unknown"}
)

print("After Filling Null Values:")
df_filled.show()
```

# **Q6. Write a query to find the total count of records for each city in a DataFrame, but only for cities where the count is greater than 100.**

**Answer:**

This query groups the dataset by `city` and counts the number of records for each city. It then filters the results to display only those cities whose record count is greater than 100.

```python
# Import required functions
from pyspark.sql.functions import count, col

# Sample Data
data = [
    ("Jaipur",),
    ("Jaipur",),
    ("Jaipur",),
    ("Lucknow",),
    ("Lucknow",),
    ("Indore",)
]

# Column name
columns = ["city"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

print("Original Data:")
df.show()

# Count records for each city
city_count = (
    df.groupBy("city")
      .agg(count("*").alias("Total_Count"))
)

# Filter cities with count greater than 100
result = city_count.filter(
    col("Total_Count") > 100
)

print("Cities with Count Greater Than 100:")
result.show()
```

# **Q7. How does the immutability of Spark DataFrames affect how you perform "data cleaning" steps like dropping columns or renaming them?**

**Answer:**

Spark DataFrames are **immutable**, which means that once a DataFrame is created, it cannot be modified directly.

When performing data cleaning operations such as dropping columns or renaming columns, Spark creates a new DataFrame instead of changing the original one.

### **Example:**

- `drop()` returns a new DataFrame after removing a column.
- `withColumnRenamed()` returns a new DataFrame with the updated column name.

### **Benefits of Immutability**

- Maintains data consistency.
- Improves fault tolerance.
- Supports parallel processing.
- Helps Spark optimize execution.

Therefore, developers usually assign the result back to a DataFrame variable.

```python
# Sample Data
data = [
    (1, "Aarav", "IT"),
    (2, "Diya", "HR"),
    (3, "Vivaan", "Finance")
]

columns = ["emp_id", "employee_name", "department"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

print("Original DataFrame:")
df.show()

# Drop department column
df_new = df.drop("department")

# Rename employee_name column
df_new = df_new.withColumnRenamed(
    "employee_name",
    "emp_name"
)

print("Updated DataFrame:")
df_new.show()
```

# **Q10. Write the code to revise a column named raw_timestamp by casting it to a TimestampType and renaming it to event_time.**

**Answer:**

This code converts the `raw_timestamp` column into `TimestampType` and then renames the column to `event_time`.

```python
# Import required functions
from pyspark.sql.functions import col
from pyspark.sql.types import TimestampType

# Sample Data
data = [
    ("2025-07-15 08:30:00",),
    ("2025-07-16 11:45:20",),
    ("2025-07-17 16:10:35",)
]

# Column name
columns = ["raw_timestamp"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

print("Original Data:")
df.show()

# Convert string column to TimestampType
df = df.withColumn(
    "raw_timestamp",
    col("raw_timestamp").cast(TimestampType())
)

# Rename column
df = df.withColumnRenamed(
    "raw_timestamp",
    "event_time"
)

print("Updated Data:")
df.show()

# Display Schema
df.printSchema()
```

# **Q11. Explain the "Shuffle" process that occurs during a grouping operation. Why is it considered a wide transformation?**

### **Answer:**

A Shuffle is the process of redistributing data across different partitions in a Spark cluster so that records with the same key can be grouped together.

Shuffle commonly occurs during operations such as:

- groupBy()
- join()
- distinct()
- reduceByKey()

### **Why Shuffle Happens**

During a grouping operation, records with the same key may exist in different partitions. Spark moves these records across partitions so that all matching keys are placed together before performing the aggregation.

### **Why is it a Wide Transformation?**

A wide transformation requires data movement between partitions.

Unlike narrow transformations, where each output partition depends on a single input partition, wide transformations require communication between multiple partitions.

### **Drawbacks of Shuffle**

- Increases network traffic.
- Requires additional memory and disk usage.
- Can slow down execution if the dataset is large.

### **Example**

The groupBy("Department") operation triggers a shuffle because Spark must bring all records of the same department into the same partition before performing aggregation.

# **Q13. How do you use the .agg() function to calculate multiple statistics at once, such as the min, max, and mean of the price column?**

### **Answer:**

The `.agg()` function in PySpark allows multiple aggregate calculations to be performed in a single operation.

In this example, the following statistics are calculated for the `price` column:

- Minimum Price
- Maximum Price
- Average Price

This helps analyze data efficiently without running separate aggregation queries.

```python
# Import required functions
from pyspark.sql.functions import min, max, avg

# Sample Data
data = [
    (120,),
    (275,),
    (430,),
    (680,),
    (950,)
]

# Column name
columns = ["price"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

print("Product Prices:")
df.show()

# Calculate minimum, maximum and average price
stats_df = df.agg(
    min("price").alias("Lowest_Price"),
    max("price").alias("Highest_Price"),
    avg("price").alias("Average_Price")
)

print("Price Statistics:")
stats_df.show()
```

# **Q14. In the context of cleaning a dataset, what is the risk of using inferSchema=true when your source data contains messy or inconsistent date formats?**

**Answer:**

When `inferSchema=true` is enabled, Spark automatically determines the data type of each column based on the input data.

If a dataset contains inconsistent or improperly formatted date values, Spark may incorrectly identify the column data type or fail to parse some records.

### **Risks**

1. Incorrect data type detection.
2. Some date values may be converted to null.
3. Filtering and sorting operations may produce inaccurate results.
4. Data quality issues can affect analysis and reporting.
5. Inconsistent formats can lead to processing errors.

### **Example**

A date column containing the following formats may create problems:

- 2025-07-15
- 15/07/2025
- July 15, 2025

### **Best Practice**

Instead of relying entirely on `inferSchema=true`, it is recommended to define the schema manually and standardize date formats before processing the data.

# **Q15. Write a final processing pipeline that:**

- Filters out duplicates.
- Fills null prices with 0.
- Groups by store_id to calculate total revenue.

### **Answer:**

This pipeline demonstrates a complete data cleaning and aggregation workflow in PySpark.

### **Steps Performed**

1. Remove duplicate records.
2. Replace null values in the `price` column with 0.
3. Group data by `store_id`.
4. Calculate total revenue for each store using the `sum()` function.

```python
# Import required function
from pyspark.sql.functions import sum

# Sample Data
data = [
    (201, 1200),
    (201, None),
    (202, 1800),
    (203, 2500),
    (202, 1800),  # Duplicate record
    (204, None)
]

# Column names
columns = ["store_id", "price"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

print("Original Data:")
df.show()

# Step 1: Remove duplicate records
df = df.dropDuplicates()

# Step 2: Fill null values in price column with 0
df = df.na.fill({"price": 0})

# Step 3: Group by store_id and calculate total revenue
final_result = (
    df.groupBy("store_id")
      .agg(sum("price").alias("Total_Revenue"))
)

print("Final Result:")
final_result.show()
```
