```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, sum, count, max, min
```

```python
spark = SparkSession.builder \
    .appName("Efficient Data Processing Assignment") \
    .master("local[*]") \
    .getOrCreate()

spark
```

```python
df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("source.csv")

df.show()
```

# **Print Schema**

The schema describes the structure of the DataFrame, including column names, data types, and whether null values are allowed. It helps verify that Spark has correctly interpreted the dataset.

```python
df.printSchema()
```

# **Display Dataset**

The show() function displays the contents of the DataFrame. It is commonly used to preview the data and verify that it has been loaded correctly.

```python
df.show()
```

# **Selecting Required Columns**

Selecting only the required columns reduces unnecessary data processing and improves query performance by reading only relevant information.

```python
df.select(
    "product_id",
    "category",
    "price",
    "status"
).show()
```

# **Filtering Records**

Filtering is used to retrieve only those records that satisfy a given condition. It helps reduce the amount of data processed.

```python
df.filter(col("amount") > 1000).show()
```

# **Renaming Columns**

Renaming columns improves readability and makes the dataset easier to understand without changing the underlying data.

```python
df = df.withColumnRenamed(
    "old_name",
    "product_name"
)

df.show()
```

# **Changing Data Types**

Data type conversion ensures that numerical columns can be used for calculations, aggregation, and other analytical operations.

```python
df = df.withColumn(
    "price",
    col("price").cast("double")
)

df.printSchema()
```

# **Adding a New Column**

A new column can be created by performing calculations on existing columns. This helps derive additional information from the dataset.

```python
df = df.withColumn(
    "final_price",
    col("base_price") * 1.18
)

df.show()
```

# **Handling NULL Values**

Missing or NULL values can affect analysis. The fillna() function replaces missing values with default values to improve data quality.

```python
from pyspark.sql.functions import when, col

df = df.withColumn(
    "user_id",
    when(col("user_id") == "NULL", "Unknown")
    .otherwise(col("user_id"))
)

df = df.fillna({
    "price": 0
})

df.show()
```

# **Filtering Processed Data**

After handling missing values, filtering is performed to extract only the records that satisfy a specific condition. This reduces the amount of data processed and improves efficiency.

```python
df.filter(col("status") == "Completed").show()
```

# **Data Processing Pipeline**

A data pipeline is a sequence of operations where data is read, transformed, filtered, and prepared for analysis. Spark executes these operations efficiently using lazy evaluation.

```python
pipeline = (
    df
    .filter(col("amount") > 1000)
    .select(
        "product_id",
        "product_name",
        "category",
        "region",
        "amount",
        "final_price"
    )
)

pipeline.show()
```

# **Wide Transformation**

A wide transformation requires data to be shuffled across different partitions. Operations such as groupBy() and join() involve data movement between executors, which may increase execution time.

```python
from pyspark.sql.functions import count, avg

df.groupBy("category").agg(
    count("*").alias("Total_Products"),
    avg("amount").alias("Average_Amount")
).show()
```

# **Lazy Evaluation**

Spark follows lazy evaluation, meaning transformations are not executed immediately. Spark waits until an action is called, allowing it to optimize the execution plan.

```python
lazy_df = df.filter(col("amount") > 1000) \
            .select("product_id", "product_name", "amount")

lazy_df.explain(True)

lazy_df.show()
```

# **Saving Data as CSV**

The processed DataFrame is saved as a CSV file so that it can be shared, analyzed, or used by other applications.

```python
pipeline.write.mode("overwrite") \
    .option("header", True) \
    .csv("output/csv_output")
```

# **Saving Data as Parquet**

Parquet is a columnar storage format that provides better compression and faster query performance compared to CSV. It is widely used in big data processing.

```python
pipeline.write.mode("overwrite") \
    .parquet("output/parquet_output")
```

# **Reading Parquet File**

The saved Parquet file is loaded back into Spark to verify that the data has been stored correctly.

```python
parquet_df = spark.read.parquet("output/parquet_output")

parquet_df.show()
```

# **Execution Plan**

The explain() method displays Spark's execution plan. It shows how Spark optimizes transformations before executing them.

```python
pipeline.explain(True)
```

# **Counting Records**

The count() action returns the total number of rows present in the DataFrame.

```python
pipeline.count()
```

# **Displaying First Five Records**

The show(5) method displays only the first five rows of the DataFrame. It is preferred over collect() for large datasets because it uses less memory.

```python
pipeline.show(5)
```
