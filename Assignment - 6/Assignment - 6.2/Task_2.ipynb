# **Q1. Explain the roles of the Driver, Cluster Manager, and Executor in a Spark application.**

### **Answer :**
Apache Spark follows a distributed computing architecture in which three main components work together to process large datasets efficiently: Driver, Cluster Manager, and Executors.

### **Driver** -

The Driver is the main controller of a Spark application. It creates the SparkSession, converts the user program into tasks, builds the Directed Acyclic Graph (DAG), schedules tasks, and sends them to the Executors. It also collects the final results after execution.

### **Cluster Manager** -

The Cluster Manager is responsible for managing the resources of the cluster. It allocates CPU cores and memory, launches Executors on worker nodes, and monitors resource utilization. Spark supports Standalone, Hadoop YARN, Kubernetes, and Mesos as Cluster Managers.

### **Executor** -

The Executor is a worker process that performs the actual data processing. It executes the tasks assigned by the Driver, processes data in parallel, stores intermediate results, and sends the final output back to the Driver.

### **Working Flow** -

Driver → Cluster Manager → Executors → Results

# **Q2. How does Spark’s Lazy Evaluation strategy improve performance when chain-processing large datasets?**

### **Answer :**

Lazy Evaluation is one of the most important features of Apache Spark. It means that Spark does not execute transformations immediately. Instead, it records all the transformations and creates a logical execution plan. The actual execution starts only when an Action such as show(), count(), collect(), or write() is called.

When multiple transformations like filter(), select(), and withColumn() are chained together, Spark combines them into a single optimized execution plan using a Directed Acyclic Graph (DAG). This optimization reduces unnecessary computations, minimizes data movement, and improves overall performance.

### **Advantages of Lazy Evaluation** -

- Optimizes the complete execution plan before running the job.
- Reduces unnecessary computations.
- Minimizes disk I/O and data shuffling.
- Improves processing speed for large datasets.
- Makes Spark more efficient than traditional data processing frameworks.

# **Q4. What is the difference between CSV and Parquet in terms of storage (row-based vs. columnar) and why does it matter for performance?**

### **Answer :**

CSV and Parquet are two widely used file formats for storing and processing data in Apache Spark. The main difference between them is the way they store data.

**CSV (Comma-Separated Values)** is a row-based file format, where data is stored row by row. It is human-readable and easy to create, but it occupies more storage space and is slower for processing large datasets because Spark has to read the entire row even if only a few columns are required.

**Parquet** is a columnar file format, where data is stored column by column. It is a compressed binary format that preserves the schema and allows Spark to read only the required columns. This significantly improves query performance and reduces memory usage.

| Feature          | CSV            | Parquet                   |
| ---------------- | -------------- | ------------------------- |
| Storage Type     | Row-based      | Columnar                  |
| File Format      | Text           | Binary                    |
| Schema           | Not Stored     | Stored                    |
| Compression      | No             | Yes                       |
| Read Performance | Slower         | Faster                    |
| Storage Size     | Larger         | Smaller                   |
| Best Use         | Small datasets | Large datasets & Big Data |


### **Why Does It Matter for Performance?**

- When working with large datasets, Spark often needs only a few columns for analysis.
- In a CSV file, Spark reads the entire row, even if only one or two columns are needed.
In a Parquet file, Spark reads only the required columns, reducing disk I/O, memory usage, and execution time.

As a result, Parquet provides better performance and is the preferred file format for big data processing in Apache Spark.

# **Q7. How does Spark use the Lineage Graph (DAG) to provide fault tolerance if a worker node fails?**

### **Answer:**

Apache Spark provides fault tolerance using a Lineage Graph, also known as a Directed Acyclic Graph (DAG). The DAG keeps track of all the transformations applied to a DataFrame or RDD from the original data source.

Whenever a transformation such as filter(), select(), or groupBy() is performed, Spark records these operations instead of immediately executing them. This sequence of operations forms the Lineage Graph.

If a worker node (Executor) fails and some data is lost, Spark does not need to restart the entire application. Instead, it uses the Lineage Graph to identify the lost partition and recomputes only the missing data by reapplying the recorded transformations. This makes Spark highly reliable and fault tolerant.

#### **Working Process**

Original Data → filter() → select() → groupBy() → Final Result

If a worker node fails during execution, Spark uses the above lineage information to recreate only the lost partition instead of processing the complete dataset again.

### **Advantages of Using DAG for Fault Tolerance**
- Recovers lost data automatically.
- Recomputes only the missing partitions.
- Avoids restarting the entire application.
- Improves reliability and fault tolerance.
- Efficiently handles failures in distributed environments.

# **Q9. Explain the concept of Predicate Pushdown in Parquet and how it affects the amount of data loaded into memory.**

### **Answer:**

Predicate Pushdown is an optimization technique used by Apache Spark while reading data from Parquet files. In this technique, Spark applies the filter condition at the file-reading level instead of loading the entire dataset into memory first.

Parquet is a columnar file format, which means data is stored column-wise. Because of this, Spark can read only the required columns and skip unnecessary data blocks that do not satisfy the filter condition.

### **How it affects memory usage:**
- Only required data is loaded into memory.
- Unnecessary rows are skipped while reading the file.
- Spark reads only the required columns from the Parquet file.
- Disk I/O is reduced.
- Memory consumption becomes lower.
- Query execution becomes faster and more efficient.

# **Q11. What is the difference between Transformations and Actions? Provide two examples of each.**

### **Answer:**

In Apache Spark, operations are classified into Transformations and Actions.

Transformations are operations that create a new DataFrame or RDD from an existing one. They are lazy, meaning Spark does not execute them immediately. Instead, Spark records these operations and executes them only when an Action is called.

Actions are operations that trigger the execution of all pending transformations and either return a result or display/save the processed data.

### **Difference between Transformations and Actions**
| Transformations               | Actions                                  |
| ----------------------------- | ---------------------------------------- |
| Create a new DataFrame or RDD | Trigger the execution of transformations |
| Lazy execution                | Immediate execution                      |
| Return another DataFrame/RDD  | Return results or display/save data      |
| Used for data processing      | Used to produce the final output         |


```python
# Example 1: Filter Transformation

filtered_df = df.filter(col("price") > 1000)

filtered_df.show()
```

# **Q12. Write the Spark command to load a Parquet file from "path/to/input", filter out any rows where user_id is null, and save the result as a CSV at "path/to/output".**

# **Q13. In Spark Architecture, what is the difference between Client Mode and Cluster Mode?**

### **Answer:**

Apache Spark supports two deployment modes: Client Mode and Cluster Mode. The main difference between them is the location where the Driver Program runs.

In **Client Mode**, the Driver runs on the client machine that submits the Spark application. The client remains connected throughout the execution and receives the output directly.

In **Cluster Mode**, the Driver runs inside the cluster on one of the worker nodes. After submitting the application, the client can disconnect because the cluster manages the entire execution.

### **Difference between Client Mode and Cluster Mode**
| **Client Mode**                                   | **Cluster Mode**                                            |
| ------------------------------------------------- | ----------------------------------------------------------- |
| Driver runs on the client machine.                | Driver runs inside the cluster.                             |
| Client must remain connected during execution.    | Client can disconnect after submitting the job.             |
| Suitable for development, testing, and debugging. | Suitable for production environments.                       |
| Less fault tolerant if the client machine fails.  | More fault tolerant because the Driver runs in the cluster. |

# **Q15. When exploring a dataset, why is it safer to use .show(5) instead of .collect() on a multi-terabyte dataset?**

### **Answer:**

When working with very large datasets, it is safer to use .show(5) instead of .collect() because .show(5) displays only the first 5 rows of the DataFrame, while .collect() retrieves all rows from the cluster to the driver program.

If the dataset is several terabytes in size, using .collect() can consume a large amount of memory and may cause the driver program to crash with an OutOfMemoryError. In contrast, .show(5) loads only a small sample of the data, making it faster and more memory-efficient.



```python
# Code Example 1: Using .show(5)
# Display only the first 5 rows
df.show(5)
```

### **Explanation:**
The show(5) method displays only the first five rows of the DataFrame. It is useful for quickly inspecting the data without loading the entire dataset into memory.

```python
# Code Example 2: Using .collect()
# Collect all rows to the driver
data = df.collect()

print(data)
```

### **Explanation:**
The collect() method retrieves all rows from the DataFrame and stores them in the driver's memory. It should be used only for small datasets because it can cause memory issues with large datasets.


| `.show(5)`                 | `.collect()`                            |
| -------------------------- | --------------------------------------- |
| Displays only a few rows   | Retrieves all rows                      |
| Uses very little memory    | Uses a large amount of memory           |
| Safe for large datasets    | Not recommended for very large datasets |
| Faster for data inspection | Slower on large datasets                |
