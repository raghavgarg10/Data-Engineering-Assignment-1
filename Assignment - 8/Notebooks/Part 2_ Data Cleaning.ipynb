## Part 2: Data Cleaning and Validation

This section performs data cleaning and validation using Pandas. Missing values, invalid emails, inconsistent formats, and referential integrity are checked. Cleaned datasets are generated and saved for the next stages of analysis.

### Step 1: Import Required Libraries

```python
# ==========================================
# Import Required Libraries
# ==========================================

import pandas as pd
import random
import re
```

### Step 2: Load Generated CSV Files

```python
# ==========================================
# Load Raw CSV Files
# ==========================================

customers = pd.read_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/customers.csv"
)

products = pd.read_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/products.csv"
)

orders = pd.read_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/orders.csv"
)

order_items = pd.read_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/order_items.csv"
)

print("All datasets loaded successfully.")
```

### Step 3 : Check Dataset Information

```python
print("Orders")
orders.info()

print("\nCustomers")
customers.info()

print("\nProducts")
products.info()

print("\nOrder Items")
order_items.info()
```

### Step 4 : Clean Orders Dataset

```python
# ==========================================
# Clean Orders Function
# ==========================================

def clean_orders(df):

    df = df.copy()

    # Convert dates
    df["order_date"] = pd.to_datetime(
        df["order_date"],
        errors="coerce",
        dayfirst=True
    )

    # Replace missing customer IDs
    df["customer_id"] = df["customer_id"].fillna("UNKNOWN")

    return df
```

```python
orders_clean = clean_orders(orders)

orders_clean.head()
```

### Step 5: Verify Cleaned Orders

```python
# Check cleaned dataset information
orders_clean.info()


# Save cleaned orders dataset
orders_clean.to_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/orders_clean.csv",
    index=False
)

print("orders_clean.csv saved successfully.")
```

### Step 6 : Clean Products Dataset

```python
# ==========================================
# Clean Products Dataset
# ==========================================

def clean_products(df):

    df = df.copy()

    # Remove extra spaces and convert product names to Title Case
    df["product_name"] = (
        df["product_name"]
        .str.strip()
        .str.title()
    )

    return df


# Call the cleaning function
products_clean = clean_products(products)

products_clean.head()
```

### Step 7: Verify Product Cleaning

```python
# Check cleaned dataset information
products_clean.info()


# Save cleaned products dataset
products_clean.to_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/products_clean.csv",
    index=False
)

print("products_clean.csv saved successfully.")
```

### Step 8 : Validate Customer Emails

```python
# ==========================================
# Validate Customer Emails
# ==========================================

def validate_emails(df):

    # Find emails that do not contain '@'
    invalid = df[
        ~df["email"].str.contains("@", na=False)
    ]

    return invalid


# Call the function
invalid_emails = validate_emails(customers)


# Display invalid email records
invalid_emails

```

### Step 9: Check Referential Integrity

```python
# ==========================================
# Check Referential Integrity
# ==========================================

def check_referential_integrity(order_items_df, orders_df):

    invalid_orders = order_items_df[
        ~order_items_df["order_id"].isin(
            orders_df["order_id"]
        )
    ]

    return invalid_orders


# Call the function
invalid_order_items = check_referential_integrity(
    order_items,
    orders
)


# Display invalid order references
invalid_order_items

invalid_order_items.info()
```

### Step 10: Data Quality Summary

```python
print("Orders:", len(orders))
print("Customers:", len(customers))
print("Products:", len(products))
print("Order Items:", len(order_items))

print("\nInvalid Emails:", len(invalid_emails))

print("Missing Customer IDs:",
      orders["customer_id"].isna().sum())

print("Negative Quantity:",
      (order_items["quantity"] < 0).sum())

print("Broken Order References:",
      len(invalid_order_items))
```

```python
# ==========================================
# Save Cleaned CSV Files
# ==========================================

customers.to_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/customers_clean.csv",
    index=False
)

products_clean.to_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/products_clean.csv",
    index=False
)

orders_clean.to_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/orders_clean.csv",
    index=False
)

order_items.to_csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/order_items_clean.csv",
    index=False
)

print("All cleaned files saved successfully!")
```
