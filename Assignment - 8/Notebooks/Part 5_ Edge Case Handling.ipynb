## Part 5: Edge Case Handling

This section tests various edge cases such as invalid order references, discount values greater than 100%, zero quantity records, and future order dates to ensure the system handles exceptional scenarios correctly.

```python
# ==========================================
# Load Cleaned Tables
# ==========================================

orders = spark.read.csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/orders_clean.csv",
    header=True,
    inferSchema=True
)

customers = spark.read.csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/customers_clean.csv",
    header=True,
    inferSchema=True
)

products = spark.read.csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/products_clean.csv",
    header=True,
    inferSchema=True
)

order_items = spark.read.csv(
    "/Volumes/workspace/default/ecommerce_analytics_system/order_items_clean.csv",
    header=True,
    inferSchema=True
)

print("All tables loaded successfully.")
```

```python
# ==========================================
# Test 1: Invalid Order ID
# ==========================================

def test_invalid_order_id():

    invalid = order_items.filter(
        ~order_items.order_id.isin(
            [row.order_id for row in orders.select("order_id").collect()]
        )
    )

    print("Invalid Order IDs:", invalid.count())

test_invalid_order_id()
```

```python
# ==========================================
# Test 2: Discount > 100
# ==========================================

def test_invalid_discount():

    invalid = order_items.filter(
        order_items.discount_percent > 100
    )

    print("Invalid Discounts:", invalid.count())

test_invalid_discount()
```

```python
# ==========================================
# Test 3: Zero Quantity
# ==========================================

def test_zero_quantity():

    invalid = order_items.filter(
        order_items.quantity == 0
    )

    print("Zero Quantity Records:", invalid.count())

test_zero_quantity()
```

```python
from pyspark.sql.functions import current_date

# ==========================================
# Test 4: Future Order Date
# ==========================================

def test_future_order_date():

    invalid = orders.filter(
        orders.order_date > current_date()
    )

    print("Future Orders:", invalid.count())

test_future_order_date()
```
